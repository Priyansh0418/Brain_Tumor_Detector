## Troubleshooting & tips
- If TensorFlow installation fails, try using the runtime's preinstalled TF (comment out the pip install above).
- Ensure your dataset contains enough images per class (dozens-minimum) for transfer learning to be effective.
- If training is slow, reduce `--epochs` or `--batch_size` to fit GPU memory.
- After you download `bt_model.h5`, place it into `models/` in your local repo and restart the backend to enable real inference in the app.
